## **Implementacion de BFS**

**Parte A:** Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el
Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

Estado | Descripcion
--|--
Estado inicial | El repartidor se encuentra en el ALMACEN
Estado objetivo | El repartidor ha llegado a ZONA NORTE
Acciones posibles | El repartidor se puede desplazar en cualquier punto cercano a el
Costo de cada acción | Consideramos el valor de 1 para cualquier tramo

**Parte B:** Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de
listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

```python
# Funcion 'bfs' que recibe el grafo, el nodo de inicio y el nodo objetivo
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    # Inicializa la cola 
    cola = [[inicio]]  
    
    # Crea un conjunto (set) vacío para registrar los nodos que ya han sido explorados
    visitados = set()  
    
    # Mientras existan caminos pendientes por explorar en la cola
    while cola:
        # Extrae el primer camino de la cola aplicando la estructura FIFO (First In, First Out)
        camino = cola.pop(0)  
        
        # Obtiene el último elemento del camino (el nodo donde nos encontramos actualmente)
        nodo_actual = camino[-1]  
        
        # Verifica si el nodo actual es la meta/objetivo buscado
        if nodo_actual == objetivo:
            return camino  # Retorna el camino completo si se encontró la solución
            
        # Evalúa si el nodo actual no ha sido visitado previamente
        if nodo_actual not in visitados:
            # Registra el nodo actual en el conjunto de visitados para evitar bucles infinitos
            visitados.add(nodo_actual)  
            
            # Recorre cada uno de los nodos adyacentes (vecinos) del nodo actual[cite: 1]
            for vecino in grafo[nodo_actual]:  
                # Crea una nueva ruta extendiendo el camino previo con el nodo vecino
                nuevo_camino = camino + [vecino]  
                
                # Agrega la nueva ruta al final de la cola para ser procesada en su turno
                cola.append(nuevo_camino)  
                
    # Si la cola se vacía sin alcanzar el objetivo, retorna None (no hay camino posible)
    return None  

# Ejecución de la función pasando el mapa, el origen y el destino
resultado = bfs(mapa, "Almacen", "ZonaNorte")

# Muestra en la pantalla la ruta encontrada  
print("Camino encontrado:", resultado)
```

## **Vectores y matrices con Numpy**

**Parte A:** Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np
obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

**Parte B:** Acceder a una la y a una columna especíca de la matriz:

```python
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```
